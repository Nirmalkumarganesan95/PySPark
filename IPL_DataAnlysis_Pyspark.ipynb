```python
# Fix: final_match was None because .show() was used during assignment in the previous cell.
# We re-assign it here correctly to resolve the AttributeError.
final_match = ipl.filter(ipl.match_id == 1426312)
first_innings_batting = final_match.filter('inning == 1')
first_innings_batting.show()
```

```python
import pyspark.sql.functions as F

# Now that first_innings_batting is a DataFrame, this aggregation will work
scorecard_df = first_innings_batting.filter("extras_type is NULL").groupBy('batter').agg(
    F.sum('batsman_runs').alias('runs'),
    F.count('ball').alias('balls'),
    F.count(F.when(F.col('batsman_runs') == 4, 1)).alias('4s'),
    F.count(F.when(F.col('batsman_runs') == 6, 1)).alias('6s'),
    F.round(F.sum('batsman_runs') * 100 / F.count('ball'), 2).alias('S/R')
)
scorecard_df.show()
```

```python
first_innings_batting.filter("batter == 'AK Markram'").show()
```

```python
batsman_order = first_innings_batting.withColumn('over-ball', (F.concat(F.col("over"), F.lit("."), F.col("ball"))).cast(FloatType())).groupBy("batter").agg(F.min("over-ball").alias("order")).orderBy("order")
batsman_order.show()
```

```python
from pyspark.sql.window import Window
import pyspark.sql.functions as F

# Using the correct Window class and orderBy method
window_spec = Window.orderBy('order')
batting_order = batsman_order.withColumn('Batting_order', F.row_number().over(window_spec))
batting_order.show()
```

```python
scorecard_df.join(batting_order, on=['batter', 'batter'], how='inner').show()
```

```python
scorecard_bowler_df = first_innings_batting.groupBy('bowler').agg(F.sum('total_runs').alias('runs_conceded'),
                                                 F.sum((F.when((F.col("extras_type") == "legbyes") | (F.col
                                                        ("extras_type") == "byes"), F.col("extra_runs")))).alias('not_by_bowler'),
                                                 F.count(F.when((F.col("extras_type").isNull()) | (F.col("extras_type") == "legbyes") | (F.col
                                                        ("extras_type") == "byes"), 1)).alias('balls'),
                                                 F.count(F.when(F.col("is_wicket") == 1,1)).alias('W'))
```

```python
import pyspark.sql.functions as F

# We perform the aggregation and transformation in one step or ensure we use the source
# to avoid "missing column" errors if the cell is re-run.
scorecard_bowler_df = first_innings_batting.groupBy('bowler').agg(
    F.sum('total_runs').alias('runs_conceded'),
    F.sum(F.when((F.col("extras_type") == "legbyes") | (F.col("extras_type") == "byes"), F.col("extra_runs"))).alias('not_by_bowler'),
    F.count(F.when((F.col("extras_type").isNull()) | (F.col("extras_type") == "legbyes") | (F.col("extras_type") == "byes"), 1)).alias('balls'),
    F.count(F.when(F.col("is_wicket") == 1, 1)).alias('W')
)

scorecard_bowler_df = scorecard_bowler_df.select(
    F.col('bowler'),
    F.concat(F.floor(F.col('balls') / 6), F.lit("."), (F.col('balls') % 6)).alias('O'),
    (F.col('runs_conceded') - F.coalesce(F.col('not_by_bowler'), F.lit(0))).alias('R'),
    F.col('W'),
    F.round((F.col('runs_conceded') / (F.col('balls') / 6)), 2).alias('Econ')
)

scorecard_bowler_df.show()
```

```python
maiden_bowler_df = first_innings_batting.groupBy('bowler', 'over').agg(F.sum('total_runs').alias('runs_conceded'),
          F.count(F.col('over')).alias('balls'),
          F.sum((F.when((F.col("extras_type") == "legbyes") | (F.col
          ("extras_type") == "byes"), F.col("extra_runs")))).alias('not_by_bowler'))
maiden_bowler_df.show()
```

```python
maiden_bowler_df = maiden_bowler_df.withColumn('runs_by_bowler', F.col('runs_conceded') - F.coalesce(F.col('not_by_bowler'),F.lit(0)))
```

```python
maiden_bowler_df.select('*').filter((F.col('runs_by_bowler') == 0) & (F.col('balls') == 6)).show()
```

```python
maiden_bowler_df = maiden_bowler_df.select('*').filter((F.col('runs_by_bowler') == 0) & (F.col('balls') == 6)).groupBy('bowler').agg(F.count('bowler').alias('M'))

maiden_bowler_df.show()
```

```python
scorecard_bowler_df.join(maiden_bowler_df, on=['bowler', 'bowler'], how='left').fillna(value=0).show()
```
