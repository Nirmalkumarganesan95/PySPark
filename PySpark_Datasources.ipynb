#PySpark Read CSV file into DataFrame


Key Points:

* PySpark supports reading a CSV file with a pipe, comma, tab, space, or any other delimiter/separator files.
* PySpark reads CSV files in parallel, leveraging multiple executor nodes to accelerate data ingestion.
* PySpark can automatically infer the schema of CSV files, eliminating the need for manual schema definition in many cases.
* Users have the flexibility to define custom schemas for CSV files, specifying data types and column names as needed.
* PySpark offers options for handling headers in CSV files, allowing users to skip headers or treat them as data rows.
* Provides robust error handling mechanisms for dealing with malformed or corrupted CSV files, ensuring data integrity.

```python
import pyspark
#1. PySpark Read CSV File into DataFrame
# Import
from pyspark.sql import SparkSession

# Create SparkSession
spark = SparkSession.builder.master("local[1]").appName("SparkByExamples.com").getOrCreate()

# Read CSV File
df = spark.read.csv("/content/zipcodes.csv")
df.printSchema()
```

##2. Reading CSV File Options

Syntax

option(self, key, value) # Using single options

options(self, **options) # Using multiple options

```python
#2.1 delimiter
#delimiter option is used to specify the column delimiter of the CSV file. By default, it is comma (,) character, but can be set to any character like pipe(|), tab (\t), space using this option.
df3 = spark.read.options(delimiter=',').csv("/content/zipcodes.csv")
df3.show(2)
```

```python
#2.2 inferSchema
#By default, the inferSchema option is set to False.
#When setting to True, Spark automatically determines column data types based on the input data.

df4 = spark.read.options(delimiter = ',', inferSchema=True).csv("zipcodes.csv")
df4.show(2)
```

```python
#Using options with key-value pair.
options = {"inferSchema": "True", "delimiter": ","}

df4 = spark.read.options(**options).csv('zipcodes.csv')
df4.show(3)
```

```python
#2.3 header
#This option is used to read the first line of the CSV file as column names.
#By default the value of this option is False , and all column types are assumed to be a string.

df5 = spark.read.options(inferSchema = True, delimiter = ',', header = True).csv('zipcodes.csv')
df5.show(3)
```

##3. Specify Custom Schema

```python
# Imports
from pyspark.sql.types import StructType,StructField, StringType, IntegerType
from pyspark.sql.types import ArrayType, DoubleType, BooleanType

# Using custom schema
schema = StructType() \
      .add("RecordNumber",IntegerType(),True) \
      .add("Zipcode",IntegerType(),True) \
      .add("ZipCodeType",StringType(),True) \
      .add("City",StringType(),True) \
      .add("State",StringType(),True) \
      .add("LocationType",StringType(),True) \
      .add("Lat",DoubleType(),True) \
      .add("Long",DoubleType(),True) \
      .add("Xaxis",IntegerType(),True) \
      .add("Yaxis",DoubleType(),True) \
      .add("Zaxis",DoubleType(),True) \
      .add("WorldRegion",StringType(),True) \
      .add("Country",StringType(),True) \
      .add("LocationText",StringType(),True) \
      .add("Location",StringType(),True) \
      .add("Decommisioned",BooleanType(),True) \
      .add("TaxReturnsFiled",StringType(),True) \
      .add("EstimatedPopulation",IntegerType(),True) \
      .add("TotalWages",IntegerType(),True) \
      .add("Notes",StringType(),True)

df_with_schema = spark.read.format("csv") \
      .option("header", True) \
      .schema(schema) \
      .load("zipcodes.csv")
df_with_schema.printSchema()
df_with_schema.show(3)
```

```python
nested_json = """[{
  "RecordNumber": 2,
  "Zipcode": 704,
  "ZipCodeType": "STANDARD",
  "City": "PASEO COSTA DEL SUR",
  "State": "PR"
},
{
  "RecordNumber": 10,
  "Zipcode": 709,
  "ZipCodeType": "STANDARD",
  "City": "BDA SAN LUIS",
  "State": "PR"
}]"""
```

```python
multiline_df = spark.read.option("multiline","true").json(spark.sparkContext.parallelize([nested_json]))
multiline_df.show()
```

### Different Ways to Handle Multiline JSON

#### 1. Reading a Multiline JSON File
If you have a file where JSON objects are formatted across multiple lines (pretty-printed), you must use the `multiline` option.

```python
# Example: Reading a multiline file
# path = 'path/to/multiline.json'
# df = spark.read.option("multiline", "true").json(path)
# df.show()
```

#### 2. Reading a Multiline String
When working with a variable containing a multiline JSON string, use the RDD approach to allow Spark to parse the string structure.

```python
# As demonstrated: using parallelize for strings
rdd = spark.sparkContext.parallelize([nested_json])
df_from_string = spark.read.option("multiline", "true").json(rdd)
df_from_string.show()
```

#### 3. Parsing JSON from a Column
You can also use the `from_json` function to extract structured data from a JSON string stored within a column of an existing DataFrame.

```python
from pyspark.sql.functions import from_json, col
from pyspark.sql.types import MapType, StringType

# Creating a dummy DataFrame with a JSON string column
data = [("1", '{"City": "SAN JUAN", "State": "PR"}')]
df_json_col = spark.createDataFrame(data, ["id", "json_data"])

# Define the schema for the JSON content
json_schema = MapType(StringType(), StringType())

# Parse the JSON column into a Map or Struct
df_parsed = df_json_col.withColumn("parsed", from_json(col("json_data"), json_schema))

df_parsed.select("id", "parsed.City", "parsed.State").show()
```

Options while writing JSON files
When writing a DataFrame to JSON files in PySpark, you can specify options to specify how you wante to write the files. Some commonly used options include:

1. path: Specifies the path where the JSON files will be saved.
2. mode: Specifies the behavior when writing to an existing directory.
3. compression: Specifies the compression codec to use when writing the JSON files (e.g., “gzip”, “snappy”).
4. dateFormat: Specifies the format for date and timestamp columns.
5. timestampFormat: Specifies the format for timestamp columns.
6. lineSep: Specifies the character sequence to use as a line separator between JSON objects.
7. encoding: Specifies the character encoding to use when writing the JSON files.

PySpark Saving modes

In PySpark, when saving DataFrames to external storage such as file systems or databases, different saving modes can be specified to control the behavior in case the target location already exists. The saving modes include: Dictionaries& Encyclopedias

1. Append: Appends the data to the existing data in the target location. If the target location does not exist, it creates a new one.
2. Overwrite: Overwrites the data in the target location if it already exists. If the target location does not exist, it creates a new one.
3. Ignore: Ignores the operation and does nothing if the target location already exists. If the target location does not exist, it creates a new one.
4. Error or ErrorIfExists: Throws an error and fails the operation if the target location already exists. This is the default behavior if no saving mode is specified.

```python
# Write with savemode example
df2.write.mode('Overwrite').json("/tmp/spark_output/zipcodes.json")
```
